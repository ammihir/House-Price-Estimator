## This is a supervised learning model to predict the values of house prices based on a bunch of parameters that we give

> We find the square of difference between actual and predicted values, these values are added and then averaged, this is the cost function. Objective is cost = 0. <br>

> We will be using **Gradient Descent** to optimize the cost function ie. reduce the cost so that it tends to 0.<br>

> **Gradient Descent** is an ensemble of decision trees!!! (a bunch of ML algos which are used together to get better results)

> **Strategy**: We randomly give weights, gradient descent iteratively tweaks the weights in an attempt to get the cost function to equal zero. <br>

> After optmizing the cost to near zero, it will throw back the weights and these weights are what will be used in the model

> **Cost Function**: An equation that tells you how wrong an estimator function is with the current weights <br>

> Hyper parameters: define how fast the model is to learn, and how much of a complicated model do you want to build

> Decision trees are implemented as 1 complex tree which defines complicated relationships, or we can have an ensemble of trees askin to asking your friends for an opinion on the same question. Gradient boosting is similar to this ensemble approach, difference being that you actually use the output of one tree as an input to the next tree <Br>

> Learn about complex trees, ensemble trees, Gradient Boosting(ensemble with sequential output used as input to next tree) Check the video in 2nd part, about Gradient Boosting

# Learnings for Feature Engineering:
Data should be 10x the number of all features.

```python
house_df.drop(['street_name','unit_number','zip_code','house_number'], axis =1, inplace = True)
```

```python
house_df
```

```python
features_df = pd.get_dummies(house_df,'garage_type','city')
```

```python
features_df.drop(['sale_price'], axis =1, inplace=True)
```

```python
X = np.array(features_df)
```

```python
y = np.array(house_df['sale_price'])
```

```python
y
```

```python
X_train, X_test, y_train, y_test =  train_test_split(X, y, test_size = 0.3)
```

```python
model = ensemble.GradientBoostingRegressor(
    n_estimators = 1000,
    learning_rate = 0.1,
    max_depth = 6,
    min_samples_leaf = 9,
    max_features = 0.1,
    loss = 'huber'
)
```

```python
model.fit(X_train,y_train)
```

```python
# Find the error rate on the training set
mse = mean_absolute_error(y_train, model.predict(X_train))
print("Training Set Mean Absolute Error: %.4f" % mse)

# Find the error rate on the test set
mse = mean_absolute_error(y_test, model.predict(X_test))
print("Test Set Mean Absolute Error: %.4f" % mse)
```

> The model was able to predict the train data with higher accuracy as compared to the test data, which is expected

# Learning for Accuracy

> Overfitting: if the training error is very low while the test error is very high <br>

> Underfitting: if the training error and test error both as high, then it is underfitted model


# Intuition:
If the model seems to overfit, then you might want to 
1. reduce the number of decision trees, n_estimators to make it generalize better or  <br>
2. make the trees less complicated

> Good fit: If the train and test error, both are low, then we have a good fit model.

n_estimators: total trees
max_depth = max decision nodes
min_samples_leaf = minimum instances to learn a decision, to avoid outliers having an impact

model = ensemble.GradientBoostingRegressor()

# Parameters we want to try
